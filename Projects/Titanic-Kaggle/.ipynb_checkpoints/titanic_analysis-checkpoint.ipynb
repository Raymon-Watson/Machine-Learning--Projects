# Titanic Analysis

This workbook provides my analysis of the Kaggle Titanic ML project (https://www.kaggle.com/competitions/titanic/overview).

The aim of this project is to get familiar with the ML workflow.

## 1. Framing the problem

The challenge of this project is to build a predictive model that answers the following question:
- "what sorts of people were more likely to survive?"

To do this, we can utilize the passenger data (including, e.g., name, age, gender, socio-econimic class, ...) to train and then test an ML model.
This suggests that the ML model we will build is:
- supervised: we have a survival metric available,
- classification: we need to classify our passengers as survived (1) or did not survive (0) in the 'Survived' column.

We begin by importing the relevant libraries for this analysis:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load Data
The dataset contained in the 'data' folder consists of 3 csv files:
- 'train.csv' - the dataset to be utilized in traininig the ML model,
- 'test.csv' - the dataset to be utilized in testing our trained ML model,
- 'gender_submission.csv' - a set of predictions that assumes all and only female passengers survive, given as an example of what a submission file should look like

## 3. Data Preparation

From the above, we have noted that there are a few columns that we can drop immediately without affecting our analysis.

Further, there are a number of categorical attributes, which should be converted to numerical values before exploring our data further. 

As a reminder this is what we decided:
- PassengerId - keep
- Survived - target variable
- Pclass - keep
- Name - Not useful, **drop**
- Sex - keep, **alter** to 0=male, 1=female
- Age - *missing* 177/891=0.199 **impute from median** as data looked skewed (don't drop, too many rows will be lost)
- SibSp - keep
- Parch - keep
- Ticket - **drop**
- Fare - keep
- Cabin - missing (687/891=0.782) **drop**
- Embarked - missing (2/891=0.002) **Impute**, convert to number (1hot) 

For now we will drop the Embarked, but later analysis we could encode this using 1hot.

For simplicity, we perform this preparation on both the training and testing data sets here, so that the testing set is prepared for our final analysis.

```python
# Drop data
df_train_prepared = df_train.drop(["Name", "Ticket","Cabin","Embarked"],axis=1)
df_test_prepared = df_test.drop(["Name", "Ticket","Cabin","Embarked"],axis=1)

# Alter sex to numerical using OrdinalEncoder

# Fit on training set:
from sklearn.preprocessing import OrdinalEncoder
df_train_cat = df_train_prepared[["Sex"]]
ordinal_encoder = OrdinalEncoder()
df_train_prepared[["Sex"]] = ordinal_encoder.fit_transform(df_train_cat)
print(f"{ordinal_encoder.categories_[0][0]}=0, {ordinal_encoder.categories_[0][1]}=1")

# Apply transform to the test set:
df_test_prepared[["Sex"]] = ordinal_encoder.transform(df_test_prepared[["Sex"]])

df_train_prepared.head()


# Impute age from median:

# First calculate the median for later reference
median = df_train_prepared[["Age"]].median()
print(f"median {median}")

# Use simple imputer to impute the median:
from sklearn.impute import SimpleImputer
age_imputer = SimpleImputer(strategy="median")
df_train_prepared[["Age"]] = age_imputer.fit_transform(df_train_prepared[["Age"]])

# Then apply this transform to the test set:
df_test_prepared[["Age"]] = age_imputer.transform(df_test_prepared[["Age"]])
```

```python
df_train_prepared.hist(bins=50, figsize=(20,15))
plt.show()
```

```python
corr_matrix = df_train_prepared.corr()
print(corr_matrix["Survived"].sort_values(ascending=False))

from pandas.plotting import scatter_matrix
scatter_matrix(df_train_prepared)
plt.show()
```
