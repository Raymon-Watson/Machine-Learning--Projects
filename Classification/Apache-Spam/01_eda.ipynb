## 2. Load and Inspect Data

The dataset consists of three folders:
- easy_ham
- hard_ham
- spam

To import the various files contained in the easy_ham and spam folders, we will utilize *Path* from the pathlib library. This provides a clean, object-oriented way to work with paths.

First, let's load in the file names of all of our spam and ham files and check how many there are:

```python
from pathlib import Path

ham_dir = Path("data/easy_ham")
spam_dir = Path("data/spam")

ham_files = list(ham_dir.iterdir())
spam_files = list(spam_dir.iterdir())

print("Ham:", len(ham_files))
print("Spam:", len(spam_files))
```

It is odd that both have 1 more than a round number. After checking through both folders, I have noticed that both contain a file named cmds, which is neither ham nor spam. I can filter out this file by noting that the rest of the files are quite long, whereas cmds contains only 4 characters.

```python
from pathlib import Path

ham_dir = Path("data/easy_ham")
spam_dir = Path("data/spam")

ham_files = list(f for f in ham_dir.iterdir() if len(f.name) > 5)
spam_files = list(f for f in spam_dir.iterdir() if len(f.name) > 5)

print("Ham:", len(ham_files))
print("Spam:", len(spam_files))
```
