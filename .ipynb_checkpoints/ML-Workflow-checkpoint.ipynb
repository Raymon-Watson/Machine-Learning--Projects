## 1. Frame the problem and look at the big picture

**Summary:** look carefully at the problem you want to solve, including current methods for solving the problem. Gather information from all relevant sources.


This is an extremely important step to start with, as we are not solely focused on building a model, instead we want to carefully select our modelling methods to serve the big picture, or find a solution to a problem that has been presented to us.

**Note on Pipelines:** pipelines is terminology for a sequence of data processing components. Typically in ML projects there is a lot of data to manipulate and many data transformations to apply, hence the overall project is constituted of various pipelines, often managed by distinct teams.

If a current solution exists, it is important to find out what it looks like, as this will give insights into potential methods which can be used to solve the problem. Overall, this initial stage is information gathering from relevant sources.

### Framing the problem
**Summary:** Explicitly frame the problem you are tackling, this will help define the method utilized in your solution.

*Framing the problem* consits of defining exactly what you are looking to do when designing your system. This is an essential first step, as correctly framing the problem means that you will not waste time applying methods that are not relevant, or answering questions that are not useful. Here, we answer questions such as:
- Is this supervised or unsupervised? Or is it reinforcement learning?
- Is this a classification task? Or a regression task? If so, what type of classification/regression task?
- Should we use batch learning or another method?
Answers to these questions should be used throughout the project to keep you on track and to define exactly what sort of ML methods are used.

### Selecting a performance measure
**Summary:** Select a relevant performance measure, commonly RMSE, but may also depend on the size of the data set and the weight you give to errors.

Next, we select a performance measure. For linear regression, this is commonly given by the *Root Mean Square Error (RMSE)*:
$$
\mathrm{RMSE}(\mathbf{X},h) = \sqrt{ \frac{1}{m} \sum_{i=1}^m \left(h(\mathbf{x}^{(i)}) - y^{(i)} \right)^2}
$$
**Note on notation:**
- $m$ is the number of instances in the dataset
- $\mathbf{x}^{(i)}$ is a vector of all the feature values of the $i^{\mathrm{th}}$ instance of the dataset, and $y^{(i)}$ is its label
- $\mathbf{X}$ is a matrix containing all feature values (excluding labels) of all instances in the dataset, where there is one row per instance, and the $i^{\mathrm{th}}$ row is equal to the transpose of $\mathbf{x}^{(i)}$
- $h$ is your system's prediction function, also called a *hypothesis*. When your system is given an instance's feature vector $\mathbf{x}^{(i)}$, it outputs a predicted value $\hat{y}^{(i)} = h(\mathbf{x}^{(i)})$
- $\mathrm{RMSE}(\mathbf{X},h)$ is the cost function measured on the set of examples using your hypothesis $h$.

Note that the RMSE is not the only performance measure for regression tasks (though it is often preferred). Another example of a performance measure is the *Mean Absolute Error (MAE)*, also called the average deviation
$$
\mathrm{MAE}(\mathbf{X},h) = \frac{1}{m} \sum_{i=1}^m \big| h(\mathbf{x}^{(i)}) - y^{(i)}\big|
$$
As a note, the above consist of two distinct ways of measuring distnaces between two vectors, and are part of the $l_k$ family of norms:
$$
||\mathbf{v}||_k = (|v_0|^k + \dots + |v_n|^k)^{1/k}
$$
The higher the norm index, the more it focuses on large values and neglects small ones, meaning e.g. that the RMSE is more sensitive to outliers than the MAE, however most data is assumed to possess Gaussian noise, for which the RMSE performs extremely well and is generally preferred.

### Check the assumptions

**Summary:** Make a list of the assumptions you are making about the data, this is effectively the stage which you double-check the overall concept you have of the project at hand.

Finally, it is good practice to list and verify the assumptions that have been made so far on the project. This is essential for making sure we catch issues at an early stage. Additionally, if this data is being fed into futher ML algorithms, perhaps by different teams, it is important to ensure that we are sending the correct data, including everything that is required in later stages. This is essentially a double-checking stage.

## 2. Get the data

**Summary:** Load the relevant data set into pandas via csv.

The first step will be to get the relevant data that we wish to analyze. This could be via download or contained in a data folder in your repository. **Important**, we will make a copy of the data before we perform our analysis, so we make sure that we always have a backup.

Typically, our data will be in csv format. We will use pandas to load the data into a DataFrame. This can be done by:
```python
import pandas as pd
df = pd.read_csv("path_to_csv")
```

### Inspect the data structure

**Summary:** use basic methods to understand the data that you have. This includes using .head(), .info(), .value_counts(), .describe(), and simple visualizations via histograms.

First, we look at the top 5 rows of our data using
```python
df.head()
```
This allows us to identify the *columns* and *grain* of our data.

Next, we can get a quick description of the data using
```python
df.info()
```
which will give us:
- total number of rows,
- each attribute's data type,
- the number of non-null values in each column.
  
The last bit of information will be very important later, as it will allow us to identify the number of columns with missing values.

Commonly, there will be columns with repeated values. Such instances may be categorical attributes, or may just be discrete values in a limited range. 
If we observed that a column "column_1" contained categorical data, we can find out exactly what categories exist and how many rows belong to each category by using .value_counts() method:
```python
df["column_1"].value_counts()
```

Looking at the other, non-categorical fields, we can use the describe() method to show a summary of their numerical attributes:
```python
df.describe()
```
This will give
- count - total number of non-null values in the column
- mean - average value of the data
- std - standard deviation
- min - minimum data value
- 25% - the value below which 25% of the data falls (i.e. 25% percentile, or first quartile)
- 50% - 50% percentile (or second quartile)
- 75% - 75% percentile (or third quartile)
- max - maximum data value



Another useful method is to plot histograms of the numerical data using hist() from Pyplot. A useful way of doing this all at once is via:
```python
df.hist(bins=50, figsize=(20,15))
plt.show()
```
which will return a histogram for each numerical attribute. From this, we can visually observe and note the trends present in the data. Some things to look out for:
- Units that the data is presented in
- Bounds on data (e.g. is there a hard limit to certain values? If so, this will need to be accounted for when modelling, since our model must account for enforced bounds.)
- Scales of units, which will be important for feature engineering (recall that it is generally practical for units to be on a similar or the same scale, as fitting to data can preference large values)
- Tails and skews of distributions, which tend to make it hard for ML algorithms to detect patterns, and can be later transformed.


### Create a test set
**Summary:** Use scikit-learn to split your data into training and testing sets at this point. Care needs to be taken to make sure these remain distinct, and are representative of the various strata of your population.


At this point, we want to split the data into training and testing with an 80/20 split (generally). To do this we will use train_test_split from sklearn.model_selection:
```python
train_set, test_set = train_test_split(data, test_size=0.2, random_state=42)
```
- data: the data you are splitting (i.e. your DataFrame)
- test_size: 80/20 split
- random_state: the seed for repeatable results (assuming the data set is at a fixed size)

#### Take care with train/test splits
If we add new data then the train/test split will change, even with a fixed seed. This means we may test data previously in the training set. The simplest solution is to **save the split**, which is viable for data sets that aren't too large.

Another way would be to use an identifier to keep the splitting stable. We won't go into much detail on this now, but come back to this later when it is useful.


Additionally, *random sampling isn't always representative.* An example of this would be sampling from a housing dataset where the neighborhood quality has a distinct distribution. If we simply randomly select, we might inadvertently bias, especially for smaller datasets. **Stratified sampling** is a solution to this, and it can be easily done using scikit-learn. If we have an important variable, "variable_1", that has a distinct stratification, we can use:
```python
train_set, test_set = train_test_split(
    data, 
    test_size=0.2, 
    random_state=42, 
    stratify=df["variable_1"]
)
```
Note that there are a variety of other ways of doing this, and we can come back to this later.


## 3. Explore the data and gain insights

Make sure to be only using the training set from here on (helpful to make a copy or save so we can experiment without losing our split). At this point we really want to take a look at various combinations of parameters and experiment with our data.

We can use scatter plots to get a better feel for our data, and observe some relations between the values making up our data set.

### Inspect correlations
We can also look at the correlation matrix, in particular we can look at how much each attribute correlates with our target variable:
```python
corr_matrix = df.corr()
corr_matrix["target_variable"].sort_values(ascending=False)
```
(Note this gives the standard correlation coefficient, also called Pearson's r). This measures *linear* correlations, and does not account for nonlinear relationships.

We can then scatter the attributes using the scatter_matrix() function from pandas. **Note:** if we have a lot of numerical attributes, it is best to focus on the few that are most highly correlated.
```python
from pandas.plotting import scatter_matrix

attributes = ["value_1", "value_2", "value_3"] # select highly correlated data

scatter_matrix(df[attributes], figsize(12, 8))
```

### Experiment with attribute combinations

At this point it may be useful to test combinations of your various attributes. A few examples would be:
- You are looking at housing in distinct districts, and have the total population and number of households. The total population by itself may not be particularly informative, but the average population per household may give insights.
- Likewise, we have the total number of rooms in the district, which may not be informative, we can instead look at the average number of rooms per household.

We can then look at the correlation matrix for these created attributes to see if we have improved our correlation over the base ones. Note that we can always come back to this later if we gain insight into a potentially useful new category.

If some data has **missing values** we can:
1. get rid of the corresponding rows
2. remove the attribute entirely
3. set the values to some fixed value (e.g. 0, the median, the mean, etc.)

These can be accomplished via:

### Text and categorical attributes

If we have text attributes with a limited number of values, we can convert them to numbers using Scikit-Learn's OrdinalEncoder:
```python
from sklearn.preprocessing import OrdinalEncoder
df_cat = df[["value_cat"]] # extract the column with text categorical data
ordinal_encoder = OrdinalEncoder() 
df_encoded = ordinal_encoder.fit_transform(df_cat) # encode the categorical data
```
We can look at the categories via:
```python
ordinal_encoder.categories_
```
**Note:** this will assume that categories close in number are close in value, which may not be correct. To fix this we can use *one-hot encoding*. This gives us an array for each category where 1 is hot and the rest are 0, i.e. cold. 
```python
from sklearn.preprocessing import OneHotEncoder
cat_encoder = OneHotEncoder()
df_cat_1hot = cat_encoder.fit_transform(df_cat)
```
This transforms from
```python
array([[0.],
       [1.],
       [2.]])
```
to
